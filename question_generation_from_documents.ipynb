```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "google/flan-t5-xl"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name).to("cuda")
```

```python
def query_llm(prompt):
    inputs = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=1024).to("cuda")
    outputs = model.generate(
        **inputs,
        max_new_tokens=300,
        do_sample=True,
        temperature=0.7,
        top_p=0.9
    )
    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
import PyPDF2

def extract_text_from_pdf(file_path):
    text = ""
    with open(file_path, "rb") as f:
        reader = PyPDF2.PdfReader(f)
        for page in reader.pages:
            page_text = page.extract_text()
            if page_text:
                text += page_text + "\n"
    return text

document_text = extract_text_from_pdf(pdf_path)
print(document_text[:1000])
```

```python
prompt_template = f"""
You are a senior compliance officer reviewing a due diligence document.

Read the following content and generate a list of 10 detailed and relevant questions
that should be asked during a due diligence process.

Use a clear numbered list. Be diverse and specific.

Document:
{document_text[:3000]}

Questions:
"""
```

```python
response = query_llm(prompt_template)
print("✅ Questions générées :\n")
print(response)
```
