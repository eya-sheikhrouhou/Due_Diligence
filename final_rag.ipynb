```python
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
import torch

# Enable 4-bit quantization to reduce GPU memory usage
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,  # Enable 4-bit mode
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True,
)

# Load tokenizer and model with 4-bit quantization
model_name = "mistralai/Mistral-7B-Instruct-v0.1"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    quantization_config=bnb_config,
    device_map="auto"
)

print("✅ Model loaded with 4-bit quantization!")
```

```python
import fitz  # PyMuPDF

def extract_text_from_pdfs(pdf_files):
    texts = []
    for file in pdf_files:
        with fitz.open(file) as doc:
            text = "\n".join([page.get_text("text") for page in doc])
            texts.append(text)
    return texts

# Extract text
texts = extract_text_from_pdfs(pdf_files)
print(f"✅ Extracted text from {len(texts)} PDFs.")
```

```python
def chunk_text(texts, chunk_size=512):
    chunks = []
    for text in texts:
        words = text.split()
        for i in range(0, len(words), chunk_size):
            chunk = " ".join(words[i:i + chunk_size])
            chunks.append(chunk)
    return chunks

chunks = chunk_text(texts)
print(f"✅ Created {len(chunks)} text chunks.")
```

```python
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer

# Load embedding model
embed_model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

# Convert text chunks into embeddings
embeddings = np.array(embed_model.encode(chunks, convert_to_numpy=True))

# Create FAISS index
index = faiss.IndexFlatL2(embeddings.shape[1])
index.add(embeddings)

print("✅ FAISS Index created!")
```

```python
def retrieve_text(query, chunks, index, top_k=1):  # Reduce from 3 to 1
    """Retrieve the most relevant text chunks using FAISS."""
    query_embedding = embed_model.encode([query], convert_to_numpy=True)
    distances, indices = index.search(query_embedding, top_k)
    retrieved_texts = [chunks[i] for i in indices[0]]
    return "\n".join(retrieved_texts)
```

```python
query = "Please provide a list of jurisdictions, other than those mentioned in 1l and 1m, where the Entity "
retrieved_text = retrieve_text(query, chunks, index)
print("Retrieved Context:\n", retrieved_text)
```

```python
def generate_rag_response(query, index, chunks):
    """Retrieve relevant context and generate a response using Mistral-7B."""
    retrieved_context = retrieve_text(query, chunks, index)
    full_query = f"Context: {retrieved_context}\n\nQuestion: {query}"

    # Tokenize input
    inputs = tokenizer(full_query, return_tensors="pt", truncation=True, max_length=2048).to("cuda")

    # Generate response (use max_new_tokens instead of max_length)
    output = model.generate(**inputs, max_new_tokens=200)

    return tokenizer.decode(output[0], skip_special_tokens=True)

# 🔹 Example usage
query = "Please provide a list of jurisdictions, other than those mentioned in 1l and 1m, where the Entity"
response = generate_rag_response(query, index, chunks)
print("Generated Answer:\n", response)
```

```python
faiss.write_index(index, "faiss_index.bin")
np.save("text_chunks.npy", np.array(chunks, dtype=object))

print("✅ FAISS index and text chunks saved!")

# Reload FAISS Index
def load_faiss_index():
    index = faiss.read_index("faiss_index.bin")
    chunks = np.load("text_chunks.npy", allow_pickle=True).tolist()
    return index, chunks

# Reload and test
index, chunks = load_faiss_index()
print("✅ FAISS index reloaded successfully!")
```
